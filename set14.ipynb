```python
# Set Unmutable Aviod Duplicates
# Set Combination Methods
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

# Union method
print("Union:", A.union(B))

# Intersection method
print("Intersection:", A.intersection(B))

# Difference method
print("Difference:", A.difference(B))

# Symmetric difference method
print("Symmetric Difference:", A.symmetric_difference(B))

# Check whether A is a subset of B
print("Is Subset:", A.issubset(B))

# Check whether A is a superset of B
print("Is Superset:", A.issuperset(B))
```

```python
#Set Operation
fruits = {"Apple", "Banana", "Mango"}

# Add one element
fruits.add("Orange")
print("After Add:", fruits)

# Add multiple elements
fruits.update(["Grapes", "Pineapple"])
print("After Update:", fruits)

# Remove an element (raises KeyError if missing)
fruits.remove("Banana")
print("After Remove:", fruits)

# Discard an element (no error if missing)
fruits.discard("Watermelon")
print("After Discard:", fruits)

# Remove and return an arbitrary element
removed = fruits.pop()
print("Removed Element:", removed)
print("After Pop:", fruits)

# Clear all elements
fruits.clear()
print("After Clear:", fruits)
```
